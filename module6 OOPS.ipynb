```python
# 10.What are class and static methods in Python?
'''Class methods
syntax is @classmethod.
are bound to the class and can modify class-level data. (operates on the class itself)
i.e, a class method can be called direcelty without need of creating an instance (object) of the class.
The first parameter of a class method is always cls.

Static methods (@staticmethod) in Python are functions inside a class that don’t depend on class or instance data.
They act like regular functions (Does not take self or cls) but are grouped in a class for organization.

 eg, under MathOperations class a static methods can organize all functions together like add or subtract to perform calculations.'''
class Car:
    wheels = 4  # Class attribute

    @classmethod
    def total_wheels(cls):
      return cls.wheels  # Accessing class attribute using 'cls'

    @staticmethod
    def car_type():
      return "This is a generic car."

print(Car.total_wheels())  # Calling the class method using the class name

my_car = Car()
print(my_car.total_wheels())  # Calling the class method using an instance


print(Car.car_type())  # Calling the static method using the class name
print(my_car.car_type()) # Calling the static method using an instance
```

```python
# 11. What is method overloading in Python?

'''Method overloading happens in same class, when multiple methods are defined with the same name but
 with different parameters/type of arguments.
i.e The method will do the same type of task, but it can handle different types of input.

Python doesn't support true method overloading like some other languages (e.g., Java),
but can simulate it using default parameters or variable arguments.'''

class Student:
    def greet(self, name=None):#default parameter (name=None) to simulate method overloading.
        if name is None:
            print("Hello!")
        else:
            print(f"Hello, {name}!")

#The greet() method in the Student class handle two different types of inputs.
obj1 = Student()

obj1.greet()

obj1.greet("Alice")  # Calling the method with an argument
```

```python
# 13. What is a property decorator in Python?
'''In Python, the @property decorator is used to define a method as a property.
Basically, it allows to make a method look like a simple attribute or variable,
but with the option to add extra behavior of getting or setting the value.

i.e it lets us access a method as if it's just a normal variable,
but still have control over what happens when require to get or set that value.'''
class Temperature:
    def __init__(self, celsius):
        self._celsius = celsius  # Private variable to store temperature in Celsius

#recommendation to make class variables private when using the @property decorator,
#so that cannot bypassing any checks or logic in your property getter or setter.

    @property
    def celsius(self):
        return self._celsius  # Access the Celsius value like an attribute
```

```python
# 2. Write a program to create an abstract class Shape with a method area(). Derive classes Circle and Rectangle
# from it and implement the area() method in both.

import abc
import math

class Shape(abc.ABC):     # Abstract base class
    @abc.abstractmethod   # Inheriting from abc module, ABC to make Shape an abstract class
    def area(self):
        pass              # Abstract method that must be implemented by subclasses

class Circle(Shape):      # Circle class
    def area(self, r):
        return math.pi * r ** 2

class Rectangle(Shape):     # Rectangle class
    def area(self, l, b):
        return l * b

cir = Circle()              #object
print(f"Area of Circle with radius 5: {round(cir.area(5),2)}")

rect = Rectangle()     #object
print(f"Area of Rectangle with length 4 and breadth 6: {rect.area(4, 6)}")

```

```python
# 5. Write a program to demonstrate encapsulation by creating a class BankAccount with private attributes
# balance and methods to deposit, withdraw, and check balance.

class BankAccount:
    def __init__(self, balance):
        self.__balance = balance  # Private attribute to store balance

    def deposit(self,amount):     # Public method to deposit money
        if amount > 0:
            self.__balance += amount
            print("Deposit successful")
        else:
            print("Invalid deposit amount")

    def withdraw(self,amount):    # Public method to withdraw money
        if amount > 0 and amount <= self.__balance:
            self.__balance -= amount
            print("Withdrawal successful")
        else:
            print("Invalid withdrawal amount or insufficient balance")

    def check_balance(self):      # Public method to check balance
        print(f"Balance: {self.__balance}")

    account= BankAccount(1000)    #object

    print("Account Information:")
    account.check_balance()

    account.deposit(500)
    account.withdraw(200)
    account.check_balance()


'''The __balance is private, so cannot directly modify or access it from outside the class.
account.__balance = 2000  # AttributeError: 'BankAccount' object has no attribute '__balance'.

Must use the public methods (deposit(), withdraw(), check_balance()) to interact with the balance.'''
```

```python
# 10. Demonstrate operator overloading by creating a class Vector and overriding the add method to add two vectors Class representing a vector.

class Vector:
    def __init__(self, x, y):
        self.x = x
        self.y = y

    def __add__(self, other):  # Add the vectors using the overloaded + operator
        return Vector(self.x + other.x, self.y + other.y)   # Make sure we're adding a Vector to another Vector

# Create two vector objects
v1 = Vector(3, 4)
v2 = Vector(1, 2)

result = v1 + v2            # This calls the __add__ method

print(f"Result of adding Vector({v1.x}, {v1.y}) and Vector({v2.x}, {v2.y}): Vector({result.x}, {result.y})")
```

```python
# 14. Create a class Employee with a method calculate_salary() that computes the salary based on hours worked
# and hourly rate. Create a derived class Manager that adds a bonus to the salary.
class Employee:
    def __init__(self, name, hours_worked, hourly_rate):
        self.name = name
        self.hours_worked = hours_worked
        self.hourly_rate = hourly_rate

    def calculate_salary(self):
        return self.hours_worked * self.hourly_rate

class Manager(Employee):
    def __init__(self, name, hours_worked, hourly_rate, bonus):
        super().__init__(name, hours_worked, hourly_rate)  # Initialize base class attributes
        self.bonus = bonus                                # Add bonus attribute


    def calculate_salary(self):
        base_salary = super().calculate_salary()  # Get the salary from Employee class
        return base_salary + self.bonus           # Add bonus for the manager

employee1 = Employee("John", 160, 25)
print(f"Employee Salary: Rs{employee1.calculate_salary()}")

manager1 = Manager("Alice", 160, 25, 5000)
print(f"Manager Salary: RS{manager1.calculate_salary()}")
```
